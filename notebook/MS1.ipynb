```python
from sklearn.preprocessing import OneHotEncoder, MinMaxScaler, StandardScaler
import category_encoders as ce
```

```python
# Define column types
cat_cols = ['Crop_Type', 'Soil_Type']  # categorical columns
num_cols = ['Soil_pH', 'Soil_Quality', 'N', 'P', 'K', 
            'Wind_Speed', 'Temperature', 'Humidity']
target_col = 'Crop_Yield'
```

```python
ohe = OneHotEncoder(drop='first', sparse_output=False)
encoded_ohe = ohe.fit_transform(df[cat_cols])

# Convert to DataFrame and join back
encoded_ohe_df = pd.DataFrame(encoded_ohe, 
                              columns=ohe.get_feature_names_out(cat_cols),
                              index=df.index)

df_ohe = pd.concat([df.drop(columns=cat_cols), encoded_ohe_df], axis=1)

print("One-Hot Encoding Applied Successfully!")
print("OHE Data Shape:", df_ohe.shape)
```

```python
df_ohe.to_csv("onehotencoding.csv" , index = False)
```

```python
from sklearn.preprocessing import MinMaxScaler

mm_scaler = MinMaxScaler()
df_mm_scaled = df_ohe.copy()
df_mm_scaled[df_mm_scaled.columns] = mm_scaler.fit_transform(df_mm_scaled[df_mm_scaled.columns])

print("MinMaxScaler applied — all features scaled to [0, 1].")
```

```python
df_mm_scaled.shape
```

```python
from sklearn.preprocessing import StandardScaler

std_scaler = StandardScaler()
df_std_scaled = df_ohe.copy()
df_std_scaled[df_std_scaled.columns] = std_scaler.fit_transform(df_std_scaled[df_std_scaled.columns])

print("StandardScaler applied — all features standardized (mean=0, std=1).")
```

```python
df_std_scaled.shape
```

```python
print("MinMax Scaled Summary:\n", df_mm_scaled[num_cols].describe().T)
print("\nStandard Scaled Summary:\n", df_std_scaled[num_cols].describe().T)
```

```python
# Convert it into csv file
df_mm_scaled.to_csv("AgriYield_OHE_MinMax.csv", index=False)
df_std_scaled.to_csv("AgriYield_OHE_Standard.csv", index=False)
```

```python
loo_encoder = ce.LeaveOneOutEncoder(cols=cat_cols)
df_loo = loo_encoder.fit_transform(df, df[target_col])

print("Leave-One-Out Encoding Applied Successfully!")
print("LOO Data Shape:", df_loo.shape)
```

```python
df_loo.to_csv("looencoding.csv" , index = False)
```

```python
mm_scaler = MinMaxScaler()
df_mm_scaled = df_loo.copy()
df_mm_scaled[num_cols + cat_cols+ [target_col]] = mm_scaler.fit_transform(df_mm_scaled[num_cols+ cat_cols+ [target_col]])

print("MinMaxScaler applied. Range now [0, 1]")
```

```python
std_scaler = StandardScaler()
df_std_scaled = df_loo.copy()
df_std_scaled[num_cols+ cat_cols+[target_col]] = std_scaler.fit_transform(df_std_scaled[num_cols+cat_cols+[target_col]])

print("StandardScaler applied. Mean≈0, Std≈1")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

features_to_check = ['N', 'P', 'K', 'Temperature', 'Humidity']

fig, axes = plt.subplots(3, len(features_to_check), figsize=(15,10))
plt.suptitle("Effect of Scaling on Numeric Features", fontsize=14)

# Original
for i, col in enumerate(features_to_check):
    sns.histplot(df[col], bins=30, color='skyblue', ax=axes[0,i])
    axes[0,i].set_title(f'Original {col}')

# MinMaxScaled
for i, col in enumerate(features_to_check):
    sns.histplot(df_mm_scaled[col], bins=30, color='limegreen', ax=axes[1,i])
    axes[1,i].set_title(f'MinMax {col}')

# StandardScaled
for i, col in enumerate(features_to_check):
    sns.histplot(df_std_scaled[col], bins=30, color='orange', ax=axes[2,i])
    axes[2,i].set_title(f'Standard {col}')

plt.tight_layout()
plt.show()
```
